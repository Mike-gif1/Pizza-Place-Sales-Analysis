## Pizza Place Sales Data Analysis

This project analyzes a full year of sales data from a fictional pizza place.  
The dataset consists of four CSV files:

- Orders  
- Order Details  
- Pizzas  
- Pizza Types  

The objective is to explore revenue performance, sales trends, customer demand patterns, and business insights using Python and data analysis techniques.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import zipfile

pd.set_option('display.max_columns', None)
```

```
zip_path = "Pizza-Place-Sales.zip"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall()

print("Files extracted successfully.")
```

```
orders = pd.read_csv("orders.csv")
order_details = pd.read_csv("order_details.csv")
pizzas = pd.read_csv("pizzas.csv")
pizza_types = pd.read_csv("pizza_types.csv")

orders.head()
```

## Data Merging

The four datasets are merged into a single dataframe to allow comprehensive analysis across orders, pizza details, pricing, and categories.


```
df = order_details.merge(pizzas, on="pizza_id", how="left")
df = df.merge(pizza_types, on="pizza_type_id", how="left")
df = df.merge(orders, on="order_id", how="left")

df.head()
```

```
df['date'] = pd.to_datetime(df['date'])
df['hour'] = pd.to_datetime(df['time']).dt.hour
df['day_of_week'] = df['date'].dt.day_name()
df['month'] = df['date'].dt.month_name()

df['revenue'] = df['quantity'] * df['price']

df.info()
```

## 1. Total Revenue

```
total_revenue = df['revenue'].sum()
print("Total Revenue: $", round(total_revenue, 2))
```

## 2. Total Quantity Sold

```
total_quantity = df['quantity'].sum()
print("Total Quantity Sold:", total_quantity)
```

## 3. Total Orders

```
total_orders = df['order_id'].nunique()
print("Total Orders:", total_orders)
```

## 4. Number of Pizza Types Sold

```
total_pizza_types = df['pizza_type_id'].nunique()
print("Number of Pizza Types Sold:", total_pizza_types)
```

## 5. Average Price of Pizzas

```
average_price = df['price'].mean()
print("Average Pizza Price: $", round(average_price, 2))
```

## 6. Peak Hours of Sales

```
hourly_sales = df.groupby('hour')['revenue'].sum()

plt.figure()
hourly_sales.plot()
plt.title("Revenue by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Revenue")
plt.show()

hourly_sales.sort_values(ascending=False).head()
```

## 7. Sales by Day of Week

```
daily_sales = df.groupby('day_of_week')['revenue'].sum()

day_order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
daily_sales = daily_sales.reindex(day_order)

plt.figure()
daily_sales.plot(kind='bar')
plt.title("Revenue by Day of Week")
plt.xlabel("Day")
plt.ylabel("Revenue")
plt.show()

daily_sales.sort_values(ascending=False)
```

## 8. Top 5 Bestselling Pizzas

```
top_5 = df.groupby('name')['quantity'].sum().sort_values(ascending=False).head(5)

plt.figure()
top_5.plot(kind='bar')
plt.title("Top 5 Bestselling Pizzas")
plt.xlabel("Pizza Name")
plt.ylabel("Quantity Sold")
plt.show()

top_5
```

## 9. Monthly Sales Trend

```
monthly_sales = df.groupby('month')['revenue'].sum()

month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

monthly_sales = monthly_sales.reindex(month_order)

plt.figure()
monthly_sales.plot()
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()

monthly_sales
```

## 10. Low Performing Pizzas

```
low_performers = df.groupby('name')['revenue'].sum().sort_values().head(5)
low_performers
```

```
category_sales = df.groupby('category')['revenue'].sum().sort_values(ascending=False)

plt.figure()
category_sales.plot(kind='bar')
plt.title("Revenue by Pizza Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()

category_sales
```

```
average_order_value = df.groupby('order_id')['revenue'].sum().mean()
print("Average Order Value: $", round(average_order_value, 2))
```
